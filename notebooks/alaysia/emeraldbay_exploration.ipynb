```python
parse = lambda s: tuple(sorted((n, float(c)) for n, c, _ in ast.literal_eval(s)))
summary_stats_df["parsed"]  = summary_stats_df.condition.map(parse)
summary_stats_df["n_drugs"] = summary_stats_df.parsed.map(len)
summary_stats_df["drugs"]   = summary_stats_df.parsed.map(lambda t: "+".join(n for n, _ in t))
summary_stats_df["doses"]   = summary_stats_df.parsed.map(lambda t: "+".join(f"{c:g}" for _, c in t))

conds = (summary_stats_df.drop_duplicates("condition")
           [["drugs", "doses", "n_drugs", "condition", "parsed"]]
           .sort_values(["n_drugs", "drugs", "doses"])
           .reset_index(drop=True))       # 93 rows

ctrl    = conds.drugs.str.startswith("DMSO")
singles = conds[(conds.n_drugs == 1) & ~ctrl]      # 44
combos  = conds[conds.n_drugs > 1]                 # 47
combos.drugs.nunique()                             # 16 distinct drug sets

# dose ladder per drug given alone
singles.groupby("drugs").doses.apply(lambda s: sorted(s, key=float))

# every (drug, dose) measured as a single agent -- the lookup M1 needs
measured = {t[0] for t in singles.parsed}
```

## combo vs. monotherapies

```python
# Combination therapies vs. their single-agent (monotherapy) arms.
# One row per unique combo condition (47). For each: which component drugs were
# also run alone, at exactly which dose, and the full mono dose ladder available.
#
# status tells you whether the Bliss additive null -- sum of single-agent log2
# growth ratios, since sensitivity is already log2(P_treated / P_control) -- is
#   complete     directly computable at the matched doses
#   dose gap     every component HAS a mono arm, but not at this dose (interpolatable)
#   no mono arm  >=1 component was never run alone at any dose (not recoverable)
g = lambda v: f"{v:g}"

mono_pairs = {t[0] for t in singles.parsed}          # (drug, dose) run as a single agent
ladder = defaultdict(list)                           # drug -> sorted mono dose ladder
for t in singles.parsed:
    drug, dose = t[0]
    ladder[drug].append(dose)
ladder = {k: sorted(v) for k, v in ladder.items()}
moa_of = dict(zip(drug_md_df.drug, drug_md_df["moa-fine"]))

def status_of(comps):
    if all((dr, ds) in mono_pairs for dr, ds in comps): return "complete"
    if all(dr in ladder for dr, _ in comps):            return "dose gap"
    return "no mono arm"

combo_tbl = pd.DataFrame([dict(
    combo  = r.drugs,
    dose   = r.doses,
    n      = r.n_drugs,
    moa    = " + ".join(dict.fromkeys(moa_of.get(dr, "?") for dr, _ in r.parsed)),
    status = status_of(r.parsed),
    # component drugs that exist as a single arm at *any* dose ([] if none)
    single_arms     = [dr for dr, _ in r.parsed if dr in ladder],
    # components whose exact (drug, dose) was also run alone ([] if none)
    exact_matches   = [f"{dr} @ {g(ds)}" for dr, ds in r.parsed if (dr, ds) in mono_pairs],
    # every dose each component was measured at as a single agent ("--" if never)
    available_doses = " | ".join(
        f"{dr}: {', '.join(g(x) for x in ladder[dr])}" if dr in ladder else f"{dr}: --"
        for dr, _ in r.parsed),
) for r in combos.itertuples()])

rank = {"complete": 0, "dose gap": 1, "no mono arm": 2}
combo_tbl = (combo_tbl.assign(_r=combo_tbl.status.map(rank))
             .sort_values(["_r", "combo", "dose"])
             .drop(columns="_r").reset_index(drop=True))

n_cells = summary_stats_df.groupby("condition").cell_line.nunique()
orphans = sorted({dr for t in combos.parsed for dr, _ in t} - set(ladder))
print(f"{len(combo_tbl)} combo conditions | {combos.drugs.nunique()} distinct drug sets | "
      f"{len(singles)} single-agent conditions over {len(ladder)} drugs")
print(f"every condition measured in {n_cells.min()}-{n_cells.max()} cell lines "
      f"(= n available to fit a per-combo interaction term)")
print(combo_tbl.status.value_counts().reindex(rank).to_string())
print(f"\nappear ONLY in combos, never alone ({len(orphans)}): {', '.join(orphans)}")

(combo_tbl.style
   .set_caption("EmeraldBay combination therapies and their single-agent coverage")
   .apply(lambda c: [{"complete":    "background-color:#d8f0dc",
                      "dose gap":    "background-color:#fdf2cf",
                      "no mono arm": "background-color:#fadbd8"}.get(v, "") for v in c],
          subset=["status"])
   .set_properties(subset=["available_doses", "moa"], **{"font-size": "90%"})
   .hide(axis="index"))
```

# Adding mechanism classes

```python
# Pharmacological design class for every combination condition (47 rows / 16 drug sets).
# This is the a-priori class -- the relationship the components were *designed* to have --
# so it comes from metadata alone and needs no growth or expression data. Use it as the
# label; derive the measured-interaction classes (synergy, transcriptional geometry)
# separately and cross-tab against it.
#
#   same-node redundancy               components converge on one protein / cascade node
#   vertical blockade                  different nodes of the SAME cascade
#   horizontal bypass                  two distinct pathways
#   same process / different mechanism  all cytotoxic, all DNA-directed, different mechanisms
#   orthogonal process                 cytotoxics acting on unrelated cell processes
#   pharmacokinetic modulation         one antiproliferative agent + partner(s) that only boost its exposure
#
# Rule-based rather than hand-labelled so every call is auditable via `basis` and so
# combos added later classify themselves. Two lookups are needed beyond drug_md_df,
# because neither distinction exists in its columns: PATHWAY places each signalling drug
# on a (pathway, cascade node), since moa-fine calls both RAF and TGFBR1 drugs kinase
# inhibitors and so cannot separate "same cascade" from "different cascade"; PROCESS gives
# each cytotoxic the process it acts on, since Gemcitabine, Oxaliplatin and Leucovorin all
# carry targets == "none" under one shared moa-fine string.

PATHWAY = {                                   # drug -> (pathway, cascade node)
    "Cetuximab": ("MAPK", "EGFR"), "Panitumumab": ("MAPK", "EGFR"),
    "Lapatinib ditosylate": ("MAPK", "EGFR/ERBB2"),
    "Trastuzumab": ("MAPK", "ERBB2"), "Pertuzumab": ("MAPK", "ERBB2"),
    "Tucatinib": ("MAPK", "ERBB2"),
    "Vociprotafib": ("MAPK", "SHP2"),
    "Adagrasib": ("MAPK", "RAS"), "Sotorasib": ("MAPK", "RAS"),
    "GDC-6036": ("MAPK", "RAS"), "RMC-6236": ("MAPK", "RAS"),
    "Dabrafenib": ("MAPK", "RAF"), "Encorafenib": ("MAPK", "RAF"),
    "Regorafenib": ("MAPK", "RAF"), "Trametinib": ("MAPK", "MEK"),
    "Galunisertib": ("TGFB", "TGFBR1"), "LY3200882": ("TGFB", "TGFBR1"),
    "PF-06952229": ("TGFB", "TGFBR1"),
}
CASCADE = ["EGFR", "ERBB2", "EGFR/ERBB2", "SHP2", "RAS", "RAF", "MEK", "TGFBR1"]

PROCESS = {                                   # cytotoxic -> process acted on
    "Capecitabine": "antimetabolite/TYMS", "Fluorouracil": "antimetabolite/TYMS",
    "Trifluridine": "antimetabolite/TYMS", "Tipiracil": "TYMP inhibition (PK)",
    "Gemcitabine": "antimetabolite/dNTP",  "Leucovorin": "folate modulation",
    "Oxaliplatin": "DNA crosslink",        "Irinotecan": "TOP1 poison",
    "Paclitaxel": "microtubule",
}
DNA_DIRECTED = {"antimetabolite/TYMS", "antimetabolite/dNTP", "folate modulation",
                "DNA crosslink", "TOP1 poison"}

# Partners that are not cytotoxic alone and are dosed to modulate the other agent:
# Leucovorin stabilises the 5-FU/TYMS ternary complex, Tipiracil blocks TYMP-mediated
# degradation of trifluridine (which is why Tipiracil's PROCESS entry is TYMP inhibition,
# not a second TYMS antimetabolite). An additive null is wrong by construction for these
# -- sub-additivity is the expected result, not a finding. Handled two ways below: a combo
# that is a single antiproliferative agent + only modulator partner(s) gets its own
# "pharmacokinetic modulation" class (design_class step 0); a modulator riding along in a
# larger multi-cytotoxic regimen (Leucovorin in FOLFOX/FOLFIRINOX) keeps the regimen's
# class but is surfaced in the modulator_partner column.
MODULATOR = {"Leucovorin", "Tipiracil"}

_tgt_md = dict(zip(drug_md_df.drug, drug_md_df.targets))

def _targets(drug):
    return {t.strip() for t in str(_tgt_md.get(drug) or "").split(",")
            if t.strip() not in ("", "none", "nan")}

def design_class(comps):
    """(class, basis) for one parsed condition -- a tuple of (drug, dose) pairs."""
    drugs = [dr for dr, _ in comps]
    tgt   = {dr: _targets(dr) for dr in drugs}
    node  = {dr: PATHWAY[dr][1] for dr in drugs if dr in PATHWAY}
    paths = {PATHWAY[dr][0] for dr in drugs if dr in PATHWAY}

    # 0. pure modulator partners -- strip them; if a single antiproliferative agent
    # remains, the combo exists to raise that agent's exposure, not to combine two
    # mechanisms (see MODULATOR). Runs first so the shared-target check below can't
    # mislabel e.g. Tipiracil+Trifluridine as redundancy on their shared TYMS annotation.
    core = [dr for dr in drugs if dr not in MODULATOR]
    if MODULATOR.intersection(drugs) and len(core) == 1:
        mods = ", ".join(sorted(MODULATOR.intersection(drugs)))
        return "pharmacokinetic modulation", f"{core[0]} boosted by {mods}"

    # 1. any two components converging on one protein or one cascade node
    for i, a in enumerate(drugs):
        for b in drugs[i + 1:]:
            shared = tgt[a] & tgt[b]
            if shared:
                return "same-node redundancy", f"shared target: {', '.join(sorted(shared))}"
            if a in node and node[a] == node.get(b):
                return "same-node redundancy", f"shared node: {node[a]}"

    # 2/3. every component is a signalling drug -> one cascade, or more than one
    if len(node) == len(drugs):
        ordered = sorted(set(node.values()), key=CASCADE.index)
        if len(paths) == 1:
            return "vertical blockade", f"{paths.pop()}: {' -> '.join(ordered)}"
        return "horizontal bypass", f"{' + '.join(sorted(paths))}: {' + '.join(ordered)}"

    # 4/5. every component is a cytotoxic -> all DNA-directed, or not
    if all(dr in PROCESS for dr in drugs):
        procs = {PROCESS[dr] for dr in drugs}
        basis = " + ".join(sorted(procs))
        return (("same process / different mechanism", basis) if procs <= DNA_DIRECTED
                else ("orthogonal process", basis))

    return "unclassified", ""

_rows = []
for r in combos.itertuples():
    cls, basis = design_class(r.parsed)
    _rows.append(dict(
        condition    = r.condition,      # raw key -- joins straight onto summary_stats_df
        drugs        = r.drugs,
        doses_uM     = r.doses,
        n_drugs      = r.n_drugs,
        design_class = cls,
        basis        = basis,
        modulator_partner = "; ".join(dr for dr, _ in r.parsed if dr in MODULATOR),
        moa          = " + ".join(dict.fromkeys(moa_of.get(dr, "?") for dr, _ in r.parsed)),
        _dose_key    = tuple(c for _, c in r.parsed),
    ))
combo_class = pd.DataFrame(_rows)

# dose_level: 1 = lowest dose combination measured for that drug set. Ranks the numeric
# dose tuple, not the doses string, so "10" doesn't sort below "5".
combo_class["dose_level"] = (combo_class.groupby("drugs")._dose_key
                             .rank(method="dense").astype(int))
combo_class = (combo_class.drop(columns="_dose_key")
               .sort_values(["design_class", "drugs", "dose_level"])
               .reset_index(drop=True)
               [["condition", "drugs", "doses_uM", "dose_level", "n_drugs",
                 "design_class", "basis", "modulator_partner", "moa"]])

assert len(combo_class) == len(combos), "lost a combo condition"
assert (combo_class.design_class != "unclassified").all(), (
    "unclassified: "
    f"{combo_class.loc[combo_class.design_class == 'unclassified', 'drugs'].unique()}")

print(f"{len(combo_class)} combo conditions | {combo_class.drugs.nunique()} drug sets "
      f"| {combo_class.design_class.nunique()} design classes")
print(combo_class.groupby("design_class")
      .agg(conditions=("condition", "size"), drug_sets=("drugs", "nunique")).to_string())

_PALETTE = {"same-node redundancy":              "background-color:#e8dff5",
            "vertical blockade":                 "background-color:#d8f0dc",
            "horizontal bypass":                 "background-color:#d6eaf8",
            "same process / different mechanism": "background-color:#fdf2cf",
            "orthogonal process":                "background-color:#fadbd8",
            "pharmacokinetic modulation":        "background-color:#e6d5c3"}
(combo_class.style
   .set_caption("EmeraldBay combination conditions by pharmacological design class")
   .apply(lambda c: [_PALETTE.get(v, "") for v in c], subset=["design_class"])
   .set_properties(subset=["condition"], **{"font-size": "80%"})
   .set_properties(subset=["basis", "moa"], **{"font-size": "90%"})
   .hide(axis="index"))
```
